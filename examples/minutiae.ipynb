# Minutiae extraction

```python
import cv2 as cv
import pyfing as pf
from pyfing.utils.notebook import show
from pyfing.utils.drawing import draw_minutiae
```

```python
# Paths
PATH_FVC = '../../datasets/'
```

```python
Example: minutiae extraction using a traditional (non-deep learning) pipeline.
```

```python
segmentation_alg = pf.Gmfs()
orientation_estimation_alg = pf.Gbfoe()
frequency_estimation_alg = pf.Xsffe()
enhancement_alg = pf.Gbfen()
minutiae_detection_alg = pf.Sbmex()

fingerprint = cv.imread(PATH_FVC + 'fvc2000/db2_b/101_1.png', cv.IMREAD_GRAYSCALE)
segmentation_mask = segmentation_alg.run(fingerprint)
orientations, _ = orientation_estimation_alg.run(fingerprint, segmentation_mask)
frequencies = frequency_estimation_alg.run(fingerprint, segmentation_mask, orientations)
enhanced_image = enhancement_alg.run(fingerprint, segmentation_mask, orientations, frequencies)
minutiae = minutiae_detection_alg.run(enhanced_image, segmentation_mask)
show(fingerprint, draw_minutiae(fingerprint, minutiae))
```

Example: end-to-end minutiae extraction using the LEADER method.

```python
fingerprint = cv.imread(PATH_FVC + 'fvc2000/db2_b/101_1.png', cv.IMREAD_GRAYSCALE)
alg = pf.Leader()
minutiae = alg.run(fingerprint)
show(fingerprint, draw_minutiae(fingerprint, minutiae))
```
